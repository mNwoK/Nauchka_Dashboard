```python
import pandas as pd
```

```python
df = pd.read_csv('teachers_dashboard_2024.csv')
df_pred = pd.read_csv('teachers_2024_attrition_predictions.csv')

df_pred['is_out_next_year'] = df_pred['risk_level'].apply(lambda x: 1 if x == 'High' else 0)
df_pred.head()
```

```python
df.update(df_pred[['is_out_next_year']])
```

```python
df_ml = pd.read_csv('teachers_ml_yearly_enriched.csv')
df['has_higher_edu'] = df['id'].map(df_ml['has_higher_edu'])
```

```python
df_ml
```

```python

df1 = pd.merge(df, df_ml[['id', 'has_higher_edu']], on='id', how='left')
df1
```
